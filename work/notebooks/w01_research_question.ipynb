## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The Question: Which pages should be reviewed first for a content refresh, expansion, or protection?

Unit of Analysis (Grain): One row per content item (a single pseudonymized page).

The Output: A ranked queue of pages, sorted by priority, including reason codes for why each page was flagged.

The Action: A human reviewer looks at the top-ranked pages and decides whether to update the content, rewrite the metadata, or merge it.

Cost of a Wrong Call: Wasted human effort. If the model ranks noise or seasonal drops at the top, reviewers waste time fixing pages that aren't actually broken, while truly declining pages lose traffic.

Why ML helps: Fixed rules (like the starter baseline) only caught about 24% of declining pages in the top 50, whereas a basic ML model caught 74%. ML can weigh multiple signals (age, position, CTR, volume) simultaneously rather than relying on rigid cutoffs.
